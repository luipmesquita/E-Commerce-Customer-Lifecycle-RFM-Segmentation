```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_raw = pd.read_excel('Online Retail.xlsx')
print(f"Sucess! Dataset was uploaded with {df_raw.shape[0]} rows and {df_raw.shape[1]} columns.")

df = df_raw.copy()

df.head()
```

```python
print(f"Entries before removing nulls: {len(df)}")
df = df.dropna(subset=['CustomerID'])
print(f"Entries after removing nulls: {len(df)}")
```

```python
df.info()
```

```python
df['CustomerID'] = df['CustomerID'].astype(int)
```

```python
df.info()
```

```python
df[(df['Quantity'] < 0)]
```

```python
# Remove negative quantities
df = df[(df['Quantity'] > 0)]
```

```python
df[(df['UnitPrice'] < 0)]
```

```python
df_valid = df
```

```python
# New column Total Revenue
df_valid['TotalRevenue'] = df_valid['Quantity'] * df_valid['UnitPrice']
```

```python
df_valid
```

```python
snapshot_date = df_valid['InvoiceDate'].max() + pd.Timedelta(days=1)
```

```python
df_valid
```

```python
snapshot_date
```

```python
# Group by CustomerID and RFM calculation
rfm = df_valid.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,  # Recency (days)
    'InvoiceNo': 'nunique',                                   # Frequency (unique bills)
    'TotalRevenue': 'sum'                                     # Monetary (€)
}).reset_index()

# Rename Columns
rfm.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']

rfm.describe()
```

```python
# Inverted Recency Scores from 1 to 5
rfm['R_Score'] = pd.qcut(rfm['Recency'], q=5, labels=[5, 4, 3, 2, 1])

# Frequency Scores, with method='first', given many clients have few purchases
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])

# Monetary Scores
rfm['M_Score'] = pd.qcut(rfm['Monetary'], q=5, labels=[1, 2, 3, 4, 5])

# Concatenation
rfm['RFM_Cell'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)
```

```python
rfm
```

```python
# Defining the segmentation of customers
def map_segment(df):
    r = int(df['R_Score'])
    f = int(df['F_Score'])
    
    if r >= 4 and f >= 4:
        return 'Champions'
    elif r >= 3 and f >= 3:
        return 'Loyal Customers'
    elif r >= 4 and f <= 2:
        return 'Recent Customers'
    elif r <= 2 and f >= 3:
        return 'At Risk'
    elif r <= 2 and f <= 2:
        return 'Lost Customers'
    else:
        return 'About to Sleep'

rfm['Segment'] = rfm.apply(map_segment, axis=1)

rfm['Segment'].value_counts()
```

```python
rfm_pivot = rfm.pivot_table(
    index='R_Score', 
    columns='F_Score', 
    values='CustomerID', 
    aggfunc='count'
)

plt.figure(figsize=(10, 6))
sns.heatmap(
    rfm_pivot, 
    annot=True,          
    fmt='d',             
    cmap='YlGnBu',     
    linewidths=0.5
)

plt.title('Customer Distribution in RxF Matrix', fontsize=14, pad=15)
plt.xlabel('Frequency Score (F)', fontsize=12)
plt.ylabel('Recency Score (R)', fontsize=12)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Histogram filtering the Top 5%
p95 = rfm['Monetary'].quantile(0.95)
sns.histplot(rfm[rfm['Monetary'] < p95]['Monetary'], bins=50, ax=axes[0], color='skyblue')
axes[0].set_title('Spending Distribution (Excluding Top 5%)')

sns.boxplot(x=rfm['Monetary'], ax=axes[1], color='lightgreen')
axes[1].set_title('Monetary dispersion with Outliers')
plt.show()
```

```python
monetary_sorted = rfm['Monetary'].sort_values(ascending=False)

cumulative_revenue = monetary_sorted.cumsum() / monetary_sorted.sum() * 100

top_20_index = int(len(monetary_sorted) * 0.2)
top_20_revenue = cumulative_revenue.iloc[top_20_index]

print(f"The top 20% of customers generated {top_20_revenue:.1f}% of total revenue.")
```
